## 🗣️ PLN: Linguagem Natural em Máquinas

**PLN (Natural Language Processing)** é a IA que entende texto e fala humana.

### Desafios do PLN 🤔

- ❓ **Ambiguidade:** "banco" = instituição ou assento?
- 😊 **Sentimentos:** "Que produto HORRÍVEL!" é positivo ou negativo?
- 🌐 **Contexto:** "não é ruim" = positivo? (negação!)
- 🗨️ **Variações:** mesma ideia, múltiplas formas

### Pipeline de PLN (Fluxo) 🔀

```
Texto Bruto
    ↓ 🔤 Tokenização (dividir em palavras)
Tokens
    ↓ ✂️ Limpeza (remover ruído)
Texto Limpo
    ↓ 📊 Vetorização (converter em números)
Números (Matriz)
    ↓ 🤖 Modelo (Naive Bayes, SVM, etc)
Predição (Classificação)
```

<summary><h1>2. 🔎 Abordagens de IA</h1></summary>

IA Simbólica versus IA Conectivista

- **IA Simbólica:** Baseada na manipulação explícita de regras, lógica e símbolos. Exemplo clássico: sistemas especialistas, que usam “se-então” para decisões estruturadas. É ideal para problemas claramente definidos, onde o conhecimento é fácil de formalizar.
- **IA Conectivista:** Inspirada no funcionamento das redes neurais biológicas, utiliza unidades simples e interconectadas (neurônios artificiais) para aprender representações complexas a partir de dados. Essas arquiteturas são a base das redes neurais artificiais e do **deep learning**.

</details>


<details>
<summary><h1>3. 📚 Tipos de Aprendizado</h1></summary>

- **Aprendizado Supervisionado:** Algoritmos aprendem a partir de exemplos rotulados, isto é, cada entrada de dados (“X”) vem acompanhada de uma resposta correta (“label”/etiqueta, “Y”). O objetivo é predizer corretamente o output para exemplos nunca vistos.
  - *Exemplo:* Classificar e-mails como “spam” ou “não-spam” a partir de histórico rotulado.
- **Aprendizado Não-supervisionado:** O algoritmo busca identificar padrões, agrupamentos ou estruturas nos dados sem saber as respostas corretas de antemão (sem labels). O clustering é o exemplo mais clássico.
  - *Exemplo:* Descobrir grupos de clientes com comportamentos parecidos em compras, sem saber previamente quem são.
- **Aprendizado por Reforço:** Um agente interage com um ambiente tomando decisões em sequência, recebendo recompensas ou punições (feedback) para aprender estratégias ótimas. É amplamente utilizado em robótica e jogos, por exemplo.
  - *Exemplo:* Treinar um robô para andar em um ambiente novo a partir de tentativas e erros.

## ✂️ SEÇÃO 1: TOKENIZAÇÃO EM PROFUNDIDADE


### O que é Tokenização?


**Tokenização** é o processo de dividir um texto em unidades menores chamadas **tokens**.


Um token pode ser:
- 🔤 **Palavra:** "python" é um token
- 🔢 **Número:** "2025" é um token
- 📍 **Pontuação:** "." ou "!" também são tokens
- 🏷️ **Emoji:** "😊" pode ser um token


Tokenização é o **primeiro passo essencial** em qualquer pipeline PLN.


### Por que é importante?


```
❌ SEM tokenização:
   Máquina vê: "O Brasil ganhou! Que jogo."
   Dificuldade: como processar texto como um tudo?


✅ COM tokenização:
   Máquina vê: ["O", "Brasil", "ganhou", "!", "Que", "jogo", "."]
   Facilidade: processa palavra por palavra
```


### Tipos de Tokenização


#### **1. Tokenização por Palavras (Word Tokenization)**


```
Entrada: "O Brasil ganhou 5-0! Que jogo incrível :)"
Saída:   ['O', 'Brasil', 'ganhou', '5-0', '!', 'Que', 'jogo', 'incrível', ':)']
```


**Desafios:**
- E-mails: "contato@empresa.com" é uma palavra ou três?
- Números: "5-0" é uma ou duas palavras?
- Emoticons: ":)" deveria ficar junto ou separado?


#### **2. Tokenização por Sentenças (Sentence Tokenization)**


```
Entrada: "O Brasil ganhou! A vitória foi inesperada. Que jogo emocionante."
Saída:   ['O Brasil ganhou!', 'A vitória foi inesperada.', 'Que jogo emocionante.']
```


**Desafios:**
- Abreviações: "Dr. Silva foi eleito." tem ponto em "Dr." e no final?
- URLs: "Veja em www.site.com.br" tem pontos que não são fim de sentença?
- Elipses: "Vamos ver... o que acontece" com "..."?


#### **3. Tokenização por Caracteres (Character Tokenization)**


```
Entrada: "IA"
Saída:   ['I', 'A']
```


Menos comum, mas útil para:
- Idiomas sem espaços (chinês, japonês)
- Análise de ortografia
- Modelos muito avançados (subpalavras)


### Código Prático 1: Tokenização Básica com NLTK

````python
# =============================================
# CÓDIGO 3: BAG OF WORDS COM SCIKIT-LEARN
# =============================================

from nltk.tokenize import word_tokenize, sent_tokenize
# Importa duas funções da biblioteca NLTK:
# - word_tokenize: divide um texto em palavras e símbolos.
# - sent_tokenize: divide um texto em sentenças (frases).

# Exemplos de textos para tokenizar
texto1 = "O Brasil ganhou 5-0! Que jogo incrível :)"
texto2 = "Comprei um celular. O preço era $499.99!"
texto3 = "E-mail: contato@empresa.com.br - Telefone: (11) 98765-4321"
# Três textos diferentes contendo palavras, números, pontuações e símbolos.
# Esses exemplos servem para demonstrar como o tokenizador lida com cada caso.

# Tokenizar palavras
tokens1 = word_tokenize(texto1, language='portuguese')
# word_tokenize() divide o texto em unidades chamadas "tokens" (palavras, números, emojis, pontuação, etc.)
# O parâmetro language='portuguese' indica que o tokenizador deve adaptar regras ao português.
print(f"Texto 1: {texto1}")
print(f"Tokens: {tokens1}")
print()

tokens2 = word_tokenize(texto2, language='portuguese')
# Mesmo processo acima, agora aplicado ao texto2.
print(f"Texto 2: {texto2}")
print(f"Tokens: {tokens2}")
print()

tokens3 = word_tokenize(texto3, language='portuguese')
# Tokenização aplicada ao texto3, contendo e-mail, telefone e símbolos.
print(f"Texto 3: {texto3}")
print(f"Tokens: {tokens3}")
print()

# Tokenizar sentenças
texto = "O Brasil ganhou! A vitória foi inesperada. Que jogo emocionante."
# Texto com múltiplas frases separadas por pontuação.
sentenças = sent_tokenize(texto, language='portuguese')
# sent_tokenize() divide um texto em sentenças completas.
# A separação é feita por pontuação (., !, ?) e regras linguísticas.

print(f"Sentencas (sent_tokenize):")
for i, sent in enumerate(sentenças, 1):
    print(f"  {i}. {sent}")
# Percorre todas as sentenças geradas e imprime cada uma numerada.
````

## 🚫 Seção 2: Remoção de Stopwords

**Stopwords** são palavras muito comuns que geralmente não contêm informação significativa.

Stopwords incluem: artigos (o, a, um), preposições (de, para, em), conjunções (e, ou, mas), etc.

### Por que remover stopwords?
- 🎯 Reduz ruído nos dados
- 📉 Diminui dimensionalidade
- ⚡ Acelera processamento
- 💡 Melhora sinais importantes

### exemplo

Exemplos: "o", "a", "de", "para", "em", "é", "e", "ou", "mas"...

```
Frase: "O gato de estimação é muito bom"
Stopwords: O, de, é, muito
Palavras-chave: gato, estimação, bom

Frase sem stopwords: "gato estimação bom"
```

### Por que remover stopwords?

✅ **Reduz dimensionalidade** (menos características = modelo mais rápido)  
✅ **Reduz ruído** (remove palavras sem informação)  
✅ **Melhora foco** (realça palavras importantes)  
✅ **Economiza espaço** (matriz de dados menor)

### ⚠️ Cuidado:
Em análise de sentimentos, remover stopwords pode perder informações críticas!
Exemplo: "não é bom" vs "é não bom" - a negação é importante!

### Código Prático 2: Remoção Básica de Stopwords

````python
# =============================================
# CÓDIGO 4: BAG OF WORDS COM SCIKIT-LEARN
# =============================================

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
# Importa:
# - stopwords: conjunto de palavras comuns (ex.: "de", "a", "que", "é") que geralmente não carregam significado importante.
# - word_tokenize: função que divide um texto em palavras e pontuações (tokens).

# Obter stopwords em português
stopwords_pt = set(stopwords.words('portuguese'))
# Cria um conjunto (set) com todas as stopwords da língua portuguesa.
# Usamos set porque busca de elementos é mais rápida que em lista.

print(f"Total de stopwords: {len(stopwords_pt)}")
# Exibe quantas stopwords existem para o português.
print(f"Exemplos: {list(stopwords_pt)[:15]}")
# Mostra as 15 primeiras stopwords só para visualização.
print()

# Reviews para testar
reviews = [
    "Adorei o produto! É de ótima qualidade.",
    "Não gostei. Produto de péssima qualidade.",
    "É ok. Preço bom mas qualidade ruim."
]
# Lista com três avaliações curtas.
# Vamos usá-las para ver o efeito de remover stopwords.

print("=" * 70)
print("ANÁLISE DE REVIEWS COM E SEM STOPWORDS")
print("=" * 70)

for i, review in enumerate(reviews, 1):

    # Tokenizar
    tokens = word_tokenize(review.lower(), language='portuguese')
    # Converte tudo pra minúsculo e divide o texto em tokens.
    # Exemplos: "produto!", "é", "ruim", "não"

    # Remover stopwords e pontuação
    tokens_filtrados = [
        token for token in tokens 
        if token not in stopwords_pt and token.isalpha()
    ]
    # Aqui criamos uma nova lista contendo apenas:
    # - tokens que NÃO são stopwords
    # - tokens que são alfabéticos (token.isalpha())
    #   → Isso remove pontuação como ".", "!", etc.
    #
    # É aqui que palavras como "não" e "mas" acabam sendo removidas,
    # porque são classificadas como stopwords.

    print(f\"\n📝 Review {i}:\")
    print(f\"   Original: {review}\")
    print(f\"   Tokens sem filtro: {tokens}\")
    print(f\"   Tokens filtrados: {tokens_filtrados}\")

# Aviso final
print(\"\\n⚠️ Observação importante:\")
print(\"   - Review 2: Perdemos a negação 'não'!\") 
print(\"   - Review 3: Perdemos a relação 'mas'!\")
# Isso é um alerta comum em PLN:
# Algumas stopwords carregam significado importante,
# principalmente para análises de sentimento.
#
# Remover "não" muda completamente o sentido de uma frase!
````

## 👜 Seção 3: BAG OF WORDS (BOW)

**Bag-of-Words (BoW)** representa documentos **contando frequências de palavras**.

Cada documento é um 'saco' de palavras, **ignorando ordem e gramática**.

### Analogia do Mundo Real 🎒

Imagine uma mochila (bag) com palavras dentro:
- Não importa a **ordem** que você coloca
- O que importa é **QUANTAS de cada tipo**

```
Frase 1: "O gato subiu no telhado"
 → Mochila: {gato: 1, subiu: 1, telhado: 1}

Frase 2: "No telhado o gato subiu"  
 → Mochila: {gato: 1, subiu: 1, telhado: 1}

⚠️ MESMA CONTAGEM! BoW não se importa com ordem!
```

### Como Funciona (Passo a Passo)

```
PASSO 1: Criar Vocabulário (lista de todas as palavras únicas)
 Documentos: "python é bom", "python é rápido"
 Vocabulário: {python, é, bom, rápido}

PASSO 2: Contar frequências para cada documento
 Doc1: python(1) é(1) bom(1) rápido(0) → [1, 1, 1, 0]
 Doc2: python(1) é(1) bom(0) rápido(1) → [1, 1, 0, 1]

PASSO 3: Matriz BoW (cada linha é um documento)
        python  é  bom  rápido
 Doc1      1    1    1     0
 Doc2      1    1    0     1
```

### Propriedades do BoW

✅ **Vantagens**:
- Super simples de entender e explicar
- Muito rápido (computação trivial)
- Funciona bem em muitas tarefas práticas
- Totalmente transparente

❌ **Limitações**:
- Perde ordem das palavras
- Perde contexto ("banco" = instituição vs "banco" = rio)
- Matriz muito esparsa (muitos zeros)
- Sem semântica ("cachorro" ≠ "cão")

### Código Prático 3: BoW Básico com scikit-learn

````python
# =============================================
# CÓDIGO 5: BAG OF WORDS COM SCIKIT-LEARN
# =============================================

from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd
# CountVectorizer: transforma textos em vetores de contagem (Bag of Words)
# pandas: usado para criar DataFrames, facilitando a visualização da matriz.

print("\n" + "=" * 70)
print("DEMONSTRAÇÃO: BAG OF WORDS")
print("=" * 70)

# Reviews de exemplo
reviews = [
    "Som excelente, muito bom, recomendo!",
    "Produto ruim, som horrível, não recomendo",
    "Bom custo-benefício, som bom",
    "Qualidade excelente, adorei",
    "Horrível, pior compra, muito ruim"
]
# Lista de textos que serão convertidos em vetores numéricos.

print("\n📝 Reviews Originais:")
for i, review in enumerate(reviews, 1):
    print(f"   {i}. {review}")
# Apenas mostra os textos para referência.

# Criar vetorizador BoW
vectorizer = CountVectorizer(lowercase=True)
# CountVectorizer:
# - Converte texto em Bag of Words.
# - lowercase=True converte tudo para minúsculas automaticamente.
# - Ele remove pontuação e tokeniza internamente.
# - Cria um VOCABULÁRIO com todas as palavras únicas.

X = vectorizer.fit_transform(reviews)
# fit_transform():
# - fit() aprende o vocabulário das reviews.
# - transform() converte cada review em um vetor de contagem.
# O resultado X é uma matriz esparsa (matriz com muitos zeros).

feature_names = vectorizer.get_feature_names_out()
# Retorna as palavras únicas do vocabulário aprendido.

print(f"\n✨ Vocabulário criado:")
print(f"   Total de palavras únicas: {len(feature_names)}")
print(f"   Palavras: {list(feature_names)}")
# Mostra quantas palavras o modelo encontrou e lista todas.

# Converter para DataFrame visual
df_bow = pd.DataFrame(
    X.toarray(),  # converte a matriz esparsa em matriz normal (array)
    columns=feature_names,  # nomes das colunas = palavras
    index=[f"Review {i+1}" for i in range(len(reviews))]  # rótulos das reviews
)
# Cria uma matriz Bag-of-Words legível:
# - Cada linha é um texto
# - Cada coluna é uma palavra do vocabulário
# - Cada célula é a quantidade de vezes que a palavra aparece na review

print("\n📊 Matriz Bag-of-Words (Contagens):")
print(df_bow)

# Análise
print("\n🔍 ANÁLISE:")
word_freq = df_bow.sum(axis=0)
# Soma cada coluna, gerando a frequência total de cada palavra no corpus.

top_words = word_freq.nlargest(5)
# Seleciona as 5 palavras mais frequentes.

print("\n📈 Top 5 palavras mais frequentes:")
for palavra, freq in top_words.items():
    print(f"   '{palavra}': {int(freq)}x")
# Exibe as palavras mais usadas e suas quantidades.

print("\n💡 Interpretação:")
print("   - Palavras como 'bom' e 'horrível' distinguem reviews")
print("   - A matriz captura sentimento sem ordem!")
# Observação importante:
# O Bag-of-Words ignora completamente a ordem das palavras.
# Apenas conta quantas vezes cada uma aparece.
# Mesmo assim, palavras como "bom", "excelente", "horrível", "ruim"
# ajudam a identificar sentimento.
````

## ⚖️ Seção 4: TF-IDF (Term Frequency-Inverse Document Frequency)

TF-IDF mede a **importância de uma palavra em um documento** dentro de uma coleção.

### Motivação: Limitações do BoW

```
Documento 1: "O gato e o gato e o gato"
Documento 2: "Machine Learning é incrível"

BoW ingênuo:
 - "o" aparece 3x em Doc1 → peso ALTO
 - "incrível" aparece 1x em Doc2 → peso BAIXO

❌ PROBLEMA: "o" é COMUM (aparece em TUDO!)
             "incrível" é RARA (aparece só em poucos docs)

✅ TF-IDF FIX: "incrível" ganharía peso MUITO maior que "o"
```

### Fórmula: TF-IDF = TF × IDF

#### **TF (Term Frequency)** - Frequência LOCAL

```
TF(palavra, doc) = Frequência em DOC / Total de palavras em DOC

Exemplo:
 Documento: "python python python java java c"
 Total: 6 palavras

 TF(python) = 3/6 = 0.5
 TF(java) = 2/6 = 0.33
 TF(c) = 1/6 = 0.17
```

#### **IDF (Inverse Document Frequency)** - Raridade GLOBAL

```
IDF(palavra) = log(Total de docs / Docs com a palavra)

Exemplo (1000 documentos):
 "o" aparece em 980/1000 docs
 IDF("o") = log(1000/980) = 0.02 → MUITO COMUM

 "Python" aparece em 50/1000 docs
 IDF("Python") = log(1000/50) = 2.996 → RARO

 "XyZaBcD" aparece em 1/1000 docs
 IDF("XyZaBcD") = log(1000/1) = 6.91 → MUITO RARO
```

#### **Resultado Final**

```
TF-IDF(palavra, doc) = TF(palavra, doc) × IDF(palavra)

Exemplo:
 TF("Python") = 0.43
 IDF("Python") = 1.61
 TF-IDF = 0.69 ← peso importante!

 TF("é") = 0.29
 IDF("é") = 0.05  
 TF-IDF = 0.015 ← peso pequeno (comum)
```

### Quando Usar TF-IDF

✅ **Busca em documentos** (Google, Wikipedia)
✅ **Recomendação de conteúdo** ("artigos similares")
✅ **Clustering de texto** (agrupar documentos parecidos)
✅ **Quando precisa capturar importância**

❌ **Quando precisa contexto profundo** (use Word Embeddings)
❌ **Quando corpus é muito pequeno** (IDF fica instável)

````python
# =============================================
# CÓDIGO 6: TF-IDF COM SCIKIT-LEARN
# =============================================

from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
# TfidfVectorizer: transforma textos em vetores numéricos usando TF-IDF.
# pandas: usado para transformar a matriz TF-IDF em tabela para visualização.

print("\n" + "=" * 70)
print("DEMONSTRAÇÃO: TF-IDF")
print("=" * 70)

corpus = [
    'Celular excelente, câmera sensacional',
    'Câmera péssima, celular ruim',
    'Celular ok, preço bom'
]
# Conjunto de textos (corpus) que serão convertidos em vetores TF-IDF.

print("\n📝 Corpus:")
for i, doc in enumerate(corpus, 1):
    print(f"   {i}. {doc}")
# Apenas exibe os documentos usados.

# Criar vetorizador TF-IDF
vectorizer = TfidfVectorizer(lowercase=True)
# TfidfVectorizer:
# 1) Converte tudo para minúsculas.
# 2) Tokeniza automaticamente.
# 3) Calcula TF-IDF:
#    - TF (Term Frequency): frequência do termo no documento.
#    - IDF (Inverse Document Frequency): importância do termo no corpus inteiro.
# 4) Palavras muito comuns recebem peso baixo.
# 5) Palavras raras recebem peso alto → são consideradas mais informativas.

X = vectorizer.fit_transform(corpus)
# fit_transform():
# - fit(): aprende o vocabulário de todas as palavras do corpus.
# - transform(): cria a matriz TF-IDF.
# O resultado é uma matriz esparsa: muitos zeros, poucos valores relevantes.

feature_names = vectorizer.get_feature_names_out()
# Lista final de palavras únicas aprendidas pelo modelo.

print(f"\n✨ Vocabulário TF-IDF:")
print(f"   Total de palavras: {len(feature_names)}")
print(f"   Palavras: {list(feature_names)}")

# Converter para DataFrame
df_tfidf = pd.DataFrame(
    X.toarray(),  # transforma matriz esparsa em matriz normal
    columns=feature_names,  # palavras como colunas
    index=[f"Doc {i+1}" for i in range(len(corpus))]  # nomes das linhas
)
# Isso cria uma tabela onde:
# - Cada linha é um documento.
# - Cada coluna é uma palavra.
# - Cada valor é o peso TF-IDF daquela palavra naquele documento.

print("\n📊 Matriz TF-IDF (Importância das Palavras):")
print(df_tfidf.round(4))
# .round(4): deixa os números mais fáceis de ler (4 casas decimais).

# Análise por documento
print("\n🏆 TOP 3 PALAVRAS MAIS IMPORTANTES POR DOCUMENTO:")
print("-" * 70)

for i in range(len(corpus)):
    # Para cada documento, pegamos os índices das 3 maiores pontuações TF-IDF.
    top_indices = X[i].toarray()[0].argsort()[-3:][::-1]
    # Explicação:
    # - argsort(): ordena os índices pelos valores.
    # - [-3:]: pega os três maiores índices.
    # - [::-1]: inverte para ficar do maior para o menor.

    print(f"\n📄 Doc {i+1}: {corpus[i][:40]}...")
    # Exibe apenas os primeiros 40 caracteres do texto.

    for rank, idx in enumerate(top_indices, 1):
        palavra = feature_names[idx]
        score = X[i].toarray()[0][idx]
        print(f"   {rank}. '{palavra}': {score:.4f}")
        # Mostra a palavra e seu peso TF-IDF.

print("\n" + "=" * 70)
print("✅ TF-IDF identifica automaticamente as palavras REALMENTE importantes!")
print("=" * 70)
````

# PARTE 6: WORD EMBEDDINGS

## 🧠 Representações Vetoriais com Significado Semântico

**Word Embeddings** representam palavras como **vetores densos** que capturam **significado semântico**.

### Diferença de Representações

```
BoW: Vetor ESPARSO (muitos zeros)
Exemplo: [1, 0, 1, 2, 0, 1, 0, ..., 0, 0, 0]
 - Dimensão: tamanho do vocabulário (~10.000)
 - Muitos zeros = desperdiça memória

Word Embeddings: Vetor DENSO (valores reais pequenos)
Exemplo: [0.25, -0.8, 0.1, ..., 0.9, -0.3]
 - Dimensão: 50-300 (MUITO menor!)
 - Todos valores significativos
 - Captura semântica!
```

### Propriedades Mágicas dos Embeddings ✨

```
Operações Vetoriais Semânticas:

"gato" ≈ "felino"
Distância entre vetores é PEQUENA (similaridade!)

"rei" - "homem" + "mulher" ≈ "rainha"
Operações aritméticas fazem sentido semântico!

"Python" ≈ "linguagem"
"Gato" ≈ "animal"
"Carro" ≈ "veículo"
```

### Técnicas de Word Embedding

1. **Word2Vec**
   - Google (2013)
   - Skip-gram e CBOW
   - Rápido e eficiente

2. **GloVe** (Global Vectors)
   - Stanford  
   - Combina frequências globais e contexto local

3. **FastText**
   - Facebook
   - Funciona com subpalavras (melhor para morfologia)

4. **ELMo**
   - Contexto profundo
   - Mesma palavra = vetores diferentes por contexto

5. **BERT/GPT**
   - Transformers modernos
   - Estado-da-arte em praticamente tudo

### Vantagens vs Limitações

✅ **Vantagens**:
- Captura relações semânticas (gato ≈ felino)
- Compacto (100-300 dims vs 10000+)
- Funciona bem com redes neurais profundas
- Transferível (treina uma vez, usa em vários problemas)

❌ **Limitações**:
- Requer mais dados para treinar bem
- Menos interpretável que BoW/TF-IDF
- Contexto estático (mesma palavra = mesmo vetor sempre)
- Requer mais computação

````python
# =============================================
# CÓDIGO 7: WORD EMBEDDINGS
# =============================================

try:
    from gensim.models import Word2Vec
    # Word2Vec: algoritmo que aprende representações vetoriais (embeddings)
    # para palavras, capturando significado e relações semânticas.

    print("\n" + "=" * 70)
    print("WORD EMBEDDINGS COM WORD2VEC")
    print("=" * 70)

    # Criar corpus de treino
    sentencas = [
        ['o', 'gato', 'preto', 'dorme'],
        ['o', 'gato', 'branco', 'corre'],
        ['o', 'cachorro', 'preto', 'corre'],
        ['gato', 'e', 'cachorro', 'são', 'animais'],
        ['python', 'é', 'linguagem', 'de', 'programação'],
        ['java', 'é', 'linguagem', 'de', 'programação'],
        ['python', 'é', 'rápido', 'e', 'poderoso'],
    ]
    # Cada lista interna é uma sentença tokenizada.
    # Word2Vec precisa de listas de palavras, não de texto puro.
    # Ele aprende padrões de contexto:
    # Ex.: "gato" aparece com "preto", "branco", "corre", "dorme"

    # Treinar modelo Word2Vec
    model = Word2Vec(
        sentences=sentencas,
        vector_size=50,  # Dimensão do embedding; vetores de 50 números
        window=3,        # Janela de contexto: quantas palavras considerar ao redor
        min_count=1,     # Palavras com frequência < 1 são ignoradas (aqui nenhuma)
        workers=1        # Número de threads usadas no treino
    )
    # Word2Vec analisa todas as sentenças e aprende vetores numéricos para cada palavra.
    # Palavras com contextos parecidos terão vetores semelhantes.

    print(f"\n📊 Informações do Modelo:")
    print(f"   Vocabulário: {len(model.wv)} palavras")
    # Mostra quantas palavras únicas foram aprendidas.

    print(f"   Dimensão de cada vetor: {model.vector_size}")
    # Mostra quantos números cada palavra possui no embedding.

    print(f"\n🧠 Exemplo de Embedding:")
    palavra = "gato"
    vetor = model.wv[palavra]
    # model.wv["gato"] retorna o vetor numérico da palavra.

    print(f"   Palavra: '{palavra}'")
    print(f"   Vetor (primeiras 10 dimensões): {vetor[:10]}")
    # Mostramos apenas as 10 primeiras das 50 posições do vetor.

    print(f"   Magnitude: {np.linalg.norm(vetor):.4f}")
    # np.linalg.norm calcula o tamanho (norma) do vetor.
    # Usado para medir a força/escala da representação.
    # (Aqui, o código assume que np foi importado — normalmente: `import numpy as np`)

    print(f"\n🔗 Palavras Similares:")
    print(f"   {'Palavra':<15} {'Similaridade':<15}")
    print("   " + "-" * 30)

    for palavra_teste in ['gato', 'python', 'programação']:
        # Para cada palavra escolhida...
        if palavra_teste in model.wv:
            similares = model.wv.most_similar(palavra_teste, topn=3)
            # most_similar encontra palavras próximas no espaço vetorial.
            # topn=3 → pegue as 3 mais parecidas.

            print(f"   {palavra_teste}:")
            for similar, score in similares:
                print(f"      {similar:<15} {score:.4f}")
                # score = similaridade de cosseno entre os vetores das palavras

except ImportError:
    print("⚠️ Gensim não instalado!")
    print("\nℹ️ Para usar Word2Vec, instale:")
    print("   !pip install gensim")
````

```
Documentos:
 Doc1: "python é bom"
 Doc2: "python é rápido"

Vocabulário = todas as palavras únicas:
 {python, é, bom, rápido}
```

### Passo 2: Codificar cada Documento

```
Doc1: "python é bom"
 → python(1) é(1) bom(1) rápido(0)
 → [1, 1, 1, 0]

Doc2: "python é rápido"
 → python(1) é(1) bom(0) rápido(1)

```

### Passo 3: Matriz BoW Final

```
       python  é  bom  rápido
Doc1      1    1    1     0
```

## 💻 EXEMPLO PRÁTICO: Reviews de Produto

Vamos classificar reviews como Positivos ou Negativos usando BoW:

```
Review 1: "Som excelente, muito bom, recomendo!"
         → Palavras importantes: excelente, bom, recomendo
         → Sentimento: POSITIVO ✅

Review 2: "Produto ruim, som horrível, não recomendo"
         → Palavras importantes: ruim, horrível
         → Sentimento: NEGATIVO ❌

         → Sentimento: POSITIVO ✅
```

Com BoW, a máquina **automaticamente** vê:

- Reviews com "ruim", "horrível" = NEGATIVO

**MAGIA**: Máquina aprendeu padrão **sem programar manualmente**!

````python
# =============================================
# CÓDIGO 8: BOW PRÁTICO COM FELIPE
# =============================================

from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

print("=" * 80)
print("🎯 DEMONSTRAÇÃO PRÁTICA: BAG-OF-WORDS")
print("=" * 80)

# Reviews reais
reviews = [
    "Som excelente, muito bom, recomendo!",
    "Produto ruim, som horrível, não recomendo",
    "Bom custo-benefício, som bom",
    "Qualidade excelente, adorei",
    "Horrível, pior compra, muito ruim"
]

labels = ["Positivo", "Negativo", "Positivo", "Positivo", "Negativo"]

print("\n📝 REVIEWS ORIGINAIS:")
print("-" * 80)
for i, (review, label) in enumerate(zip(reviews, labels), 1):
    emoji = "😊" if label == "Positivo" else "😞"
    print(f"{i}. {emoji} [{label}] {review}")

# Criar BoW
vectorizer = CountVectorizer(lowercase=True)
X = vectorizer.fit_transform(reviews)
feature_names = vectorizer.get_feature_names_out()

print(f"\n✨ VOCABULÁRIO CRIADO:")
print(f"   Total de palavras únicas: {len(feature_names)}")
print(f"   Palavras: {sorted(list(feature_names))}")

# DataFrame visual
df_bow = pd.DataFrame(
    X.toarray(),
    columns=feature_names,
    index=[f"Review {i}" for i in range(len(reviews))]
)

print("\n📊 MATRIZ BAG-OF-WORDS:")
print(df_bow)

# Análise
print("\n🔍 ANÁLISE PROFUNDA:")
print("-" * 80)

word_freq = df_bow.sum(axis=0)
top_words = word_freq.nlargest(5)

print(f"\n📈 TOP 5 PALAVRAS MAS FREQUENTES:")
for palavra, freq in top_words.items():
    print(f"   '{palavra}': aparece {int(freq)}x")

print("\n✅ CONCLUSÃO:")
print("   - BoW captura automaticamente palavras discriminativas")
print("   - 'bom', 'excelente', 'adorei' → sentimento POSITIVO")
print("   - 'ruim', 'horrível' → sentimento NEGATIVO")
print("   - Máquina APRENDEU sem programação manual! 🤖
````

# ⚖️ 8: TF-IDF - BOW COM INTELIGÊNCIA

##  TF-IDF Prático com Felipe

**Apresentado por:** Felipe  
**Tópico:** Representações de Texto - Evoluindo BoW  
**Objetivo:** Entender por que algumas palavras são mais importantes que outras

---

## O Problema do BoW Simples

No BoW clássico, **todas as palavras tem peso igual**:

```
Documento 1: "O gato e o gato e o gato"
Documento 2: "Machine Learning é incrível"

BoW ingênuo vê:
 - "o" aparece 3x em Doc1 → peso ALTO
 - "incrível" aparece 1x em Doc2 → peso BAIXO

❌ PROBLEMA: "o" é PALAVRA COMUM (aparece em TUDO!)
             "incrível" é RARA mas IMPORTANTE!

✅ SERIA MELHOR: "incrível" ter peso muito MAIOR que "o"
```

## A Solução: TF-IDF

**TF-IDF** (Term Frequency - Inverse Document Frequency) é como BoW aprendeu a ser inteligente!

### Equação Mágica

```
TF-IDF(palavra, doc) = TF(palavra, doc) × IDF(palavra)

Onde:
- TF = "Quão frequente é AQUI?" (no documento)
- IDF = "Quão rara é NO GERAL?" (no corpus inteiro)
```

### Exemplo Prático

```
Documento: "Python é ótimo. Python é rápido. Python."
Corpus: 100 documentos

Para "Python":
 TF = 3/7 = 0.43 (aparece 3x em 7 palavras)
 IDF = log(100/20) = 1.61 (aparece em 20/100 docs)
 TF-IDF = 0.69 ← IMPORTANTE!

Para "é":
 TF = 2/7 = 0.29 (aparece 2x em 7 palavras)
 IDF = log(100/95) = 0.05 (aparece em 95/100 docs)
 TF-IDF = 0.015 ← POUCO IMPORTANTE

Resultado: "Python" tem peso 46x MAIOR que "é"! 🎯
```

## Por que TF-IDF é Melhor?

| Aspecto | BoW | TF-IDF |
|---------|-----|--------|
| Peso de palavras | Iguais | Diferenciado |
| Palavras comuns | Alto | Baixo |
| Palavras raras | Baixo | Alto |
| Captura importância | Não | Sim |
| Busca em Wikipedia | ❌ Vê "o" como importante | ✅ Encontra palavras-chave |

````python
# =============================================
# CÓDIGO 9: TF-IDF COM FELIPE
# =============================================

from sklearn.feature_extraction.text import TfidfVectorizer
import pandas as pd
import numpy as np

print("=" * 80)
print("⚖️ DEMONSTRAÇÃO: TF-IDF vs BOW")
print("=" * 80)

corpus = [
    'Celular excelente, câmera sensacional',
    'Câmera péssima, celular ruim',
    'Celular ok, preço bom'
]

print("\n📝 CORPUS:")
for i, doc in enumerate(corpus, 1):
    print(f"   {i}. {doc}")

# Criar TF-IDF
vectorizer_tfidf = TfidfVectorizer(lowercase=True)
X_tfidf = vectorizer_tfidf.fit_transform(corpus)
feature_names = vectorizer_tfidf.get_feature_names_out()

# Converter para DataFrame
df_tfidf = pd.DataFrame(
    X_tfidf.toarray(),
    columns=feature_names,
    index=[f"Doc {i+1}" for i in range(len(corpus))]
)

print("\n📊 MATRIZ TF-IDF (Importância das Palavras):")
print(df_tfidf.round(4))

# Análise detalhada
print("\n🏆 TOP 3 PALAVRAS MAIS IMPORTANTES POR DOCUMENTO:")
print("-" * 80)

for i in range(len(corpus)):
    top_indices = X_tfidf[i].toarray()[0].argsort()[-3:][::-1]

    print(f"\n📄 Doc {i+1}: {corpus[i][:40]}...")
    print(f"   Top 3:")
    for rank, idx in enumerate(top_indices, 1):
        palavra = feature_names[idx]
        score = X_tfidf[i].toarray()[0][idx]
        print(f"      {rank}. '{palavra}': {score:.4f}")

print("\n" + "=" * 80)
print("✅ TF-IDF identifica automaticamente as palavras REALMENTE importantes!")
print("=" * 80)
````

**Grande pergunta:** Como transformar números em **DECISÕES**?

## 🧠 Classificação: O Que É?

### Definição Simples

**Classificação** é prever uma **categoria** baseada em **características de entrada**.

### Exemplos Reais

```
1️⃣ Email → [SPAM] ou [LEGÍTIMO]
2️⃣ Review → [POSITIVO] ou [NEGATIVO] ou [NEUTRO]
```

```
CLASSIFICAÇÃO:
 Output: Categoria (valores discretos)
 Exemplo: Spam/Legítimo, Sim/Não
 Métrica: Acurácia, Precision, Recall

REGRESSÃO:
 Output: Número contínuo
 Exemplo: Preço, Temperatura
 Métrica: MAE, RMSE

Nesta parte: CLASSIFICAÇÃO com Naive Bayes!
```

---

## 🎯 Naive Bayes: O Classificador Ingênuo

### O Nome Vem Daonde?

**"Naive"** (ingênuo) porque **assume que todas as características são independentes**.

```
Suposição Naive (simplificação drástica!):
 P(palavra1 E palavra2 | SPAM) = P(palavra1|SPAM) × P(palavra2|SPAM)

Realidade (muito mais complexa):
 Essas probabilidades são CORRELACIONADAS
 Mas o modelo IGNORA isso!

Paradoxo: Apesar de ingênuo, funciona MUITO BEM na prática! 🤔
```

### Como Funciona: Exemplo Prático

```
INPUT: Email com palavras: ["ganhe", "dinheiro", "agora"]

PASSO 1: Calcular P(SPAM) e P(LEGÍTIMO)
 Emails de treino: 100 totais (30 SPAM, 70 LEGÍTIMO)
 P(SPAM) = 30/100 = 0.3
 P(LEGÍTIMO) = 70/100 = 0.7

PASSO 2: P(palavra | SPAM)
 Nos 30 emails SPAM:
 - "ganhe" aparece em 25 → P(ganhe|SPAM) = 0.83
 - "dinheiro" aparece em 24 → P(dinheiro|SPAM) = 0.80
 - "agora" aparece em 20 → P(agora|SPAM) = 0.67

PASSO 3: P(palavra | LEGÍTIMO)
 Nos 70 emails LEGÍTIMO:
 - "ganhe" aparece em 2 → P(ganhe|LEGÍTIMO) = 0.03
 - "dinheiro" aparece em 3 → P(dinheiro|LEGÍTIMO) = 0.04
 - "agora" aparece em 10 → P(agora|LEGÍTIMO) = 0.14

PASSO 4: Teorema de Bayes (A MÁGICA!)
 P(SPAM | palavras) ∝ 0.83 × 0.80 × 0.67 × 0.3 ≈ 0.133
 P(LEGÍTIMO | palavras) ∝ 0.03 × 0.04 × 0.14 × 0.7 ≈ 0.000012

PASSO 5: Comparar e Decidir
 0.133 > 0.000012
 PREDIÇÃO: SPAM ✅
```

### Vantagens do Naive Bayes

✅ **Rápido demais** - cálculos muito simples
✅ **Pouco dados** - funciona bem com poucos exemplos
✅ **Interpretável** - vemos quais palavras influenciam
✅ **Bom baseline** - começar por este, depois melhorar

### Limitações

❌ Suposição de independência é FALSA (mas funciona!)
❌ Sensível a palavras não-vistas no treino
❌ Menos preciso que modelos mais sofisticados
❌ Não captura interações entre features

````python
# =============================================
# CÓDIGO 10: NAIVE BAYES COM FELIPE
# =============================================

from sklearn.naive_bayes import MultinomialNB
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.model_selection import train_test_split
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
import pandas as pd

print("=" * 80)
print("🤖 DEMONSTRAÇÃO: NAIVE BAYES PARA CLASSIFICAÇÃO")
print("=" * 80)

# Dataset
reviews = [
    ("Adorei o produto! Excelente qualidade!", "Positivo"),
    ("Produto muito bom, recomendo!", "Positivo"),
    ("Não gostei. Produto péssimo.", "Negativo"),
    ("Horrível, pior compra da minha vida", "Negativo"),
    ("Qualidade excelente, muito bom!", "Positivo"),
    ("Decepcionante, não recomendo", "Negativo"),
]

textos = [r[0] for r in reviews]
labels = [r[1] for r in reviews]

print("\n📚 DATASET DE TREINO:")
for texto, label in reviews:
    emoji = "😊" if label == "Positivo" else "😞"
    print(f"   {emoji} [{label}] {texto}")

# Vetorizar com BoW
vectorizer = CountVectorizer(lowercase=True)
X = vectorizer.fit_transform(textos)

# Dividir em treino/teste
X_train, X_test, y_train, y_test = train_test_split(
    X, labels, test_size=0.33, random_state=42
)

# Treinar Naive Bayes
modelo_nb = MultinomialNB()
modelo_nb.fit(X_train, y_train)

print("\n✅ Modelo treinado!")

# Prever
y_pred_train = modelo_nb.predict(X_train)
y_pred_test = modelo_nb.predict(X_test)

acc_train = accuracy_score(y_train, y_pred_train)
acc_test = accuracy_score(y_test, y_pred_test)

print(f"\n📊 ACURÁCIA:")
print(f"   Treino: {acc_train:.2%}")
print(f"   Teste: {acc_test:.2%}")

# Testar em novos exemplos
print("\n🧪 TESTANDO EM NOVOS TEXTOS:")
novos_textos = [
    "Produto maravilhoso!",
    "Que decepção, muito ruim",
]

for novo_texto in novos_textos:
    X_novo = vectorizer.transform([novo_texto])
    predicao = modelo_nb.predict(X_novo)[0]
    probabilidade = modelo_nb.predict_proba(X_novo)[0]

    print(f"\n   Texto: '{novo_texto}'")
    print(f"   Predição: {predicao}")
    print(f"   Probabilidades: Positivo={probabilidade[1]:.2%}, Negativo={probabilidade[0]:.2%}")
````

## 🧠 Do Linear para Logístico

### O Problema com Regressão Linear

```
Regressão Linear: y = wx + b

Problema: Saída é CONTÍNUA
 - Para email: pode retornar -5, 10, 1000, ...
 - Queremos probabilidade entre 0 e 1!

❌ y = -5 significa -500% de chance? IMPOSSÍVEL!
❌ y = 10 significa 1000% de chance? NÃO FAZ SENTIDO!

Solução: Usar função que força output entre 0 e 1
```

### Função Sigmoid (A Mágica!)

```
sigmoid(z) = 1 / (1 + e^(-z))

Propriedades:
 - Input: QUALQUER número real (-∞ até +∞)
 - Output: SEMPRE entre 0 e 1 ✅
 - Curva suave e diferenciável ✅

Valores:
 z → -∞ : sigmoid → 0 (certeza de NEGATIVO)
 z = 0 : sigmoid = 0.5 (50-50)
 z → +∞ : sigmoid → 1 (certeza de POSITIVO)
```

### Logistic Regression = Linear + Sigmoid

```
Passo 1: Combinação LINEAR das features
 z = w₁×x₁ + w₂×x₂ + ... + wₙ×xₙ + b

Passo 2: Aplicar SIGMOID
 P(y=1|x) = sigmoid(z) = 1/(1 + e^(-z))

Resultado: Probabilidade entre 0 e 1!
```

---

## 💡 Aprendizado: Gradient Descent

Como ajustar w e b para melhorar?

### Loss Function

```
Cross-Entropy Loss:
Loss = -y × log(ŷ) - (1-y) × log(1-ŷ)

Interpretação:
 - Se y=1 e ŷ=1: Loss ≈ 0 (ÓTIMO!)
 - Se y=1 e ŷ=0: Loss → ∞ (HORRÍVEL!)
 - Se y=0 e ŷ=0: Loss ≈ 0 (ÓTIMO!)
```

### Gradient Descent (Otimizador)

```
Algoritmo:
 1. Começar com w e b aleatórios
 2. Calcular Loss no treino
 3. Calcular gradiente: ∂Loss/∂w, ∂Loss/∂b
 4. Atualizar: w ← w - learning_rate × ∂Loss/∂w
 5. Repetir até convergência

Cada iteração: dados forçam modelo a melhorar!
```

---

✅ Não assume independência (diferente de Naive Bayes!)
✅ Aprendizado baseado em otimização
✅ Probabilidades bem calibradas  
✅ Interpretável (sabemos o peso de cada feature)

❌ Fronteira LINEAR (limitação!)
❌ Não captura interações complexas
→ Solução: Support Vector Machine!
