# PARTE 1: O que é Inteligência Artificial?

## 🧠 Definição

A **Inteligência Artificial (IA)** é um ramo da computação dedicado a criar sistemas capazes de simular capacidades típicas dos seres humanos, como raciocínio lógico, percepção sensorial, aprendizado, compreensão da linguagem natural e tomada de decisão.

**IA** é a capacidade de máquinas **aprender e tomar decisões** sem serem explicitamente programadas para cada situação.

## ⏳ Histórico

```
1950s ─► "Posso fazer uma máquina pensar?" (Turing)
1956  ─► O termo "Inteligência Artificial" é cunhado (conferência de Dartmouth)
1970s ─► Sistemas especialistas (regras manuais)
1990s ─► Deep Blue vence Kasparov no xadrez
2010s ─► Deep Learning revoluciona tudo
2020s ─► ChatGPT, BERT, transformers dominam
```

O termo \"Inteligência Artificial\" foi cunhado em 1956, durante a famosa conferência de Dartmouth. Alan Turing, nos anos 1950, propôs o célebre **Teste de Turing**, sugerindo um critério para definir se uma máquina pode ser considerada inteligente.

## 🎯 Tipos de IA

```
🔴 IA Fraca: faz uma tarefa bem (ex: xadrez, reconhecimento facial)
🟢 IA Forte: poderia fazer qualquer tarefa cognitiva (ficção científica)
💙 IA Geral: inteligência humana em máquina (ainda é sonho)
```

# PARTE 2: Abordagens de IA

## 🔎 IA Simbólica vs IA Conectivista

### IA Simbólica
Baseada na manipulação explícita de **regras, lógica e símbolos**.

- Exemplo clássico: sistemas especialistas ("se-então")
- Ideal para problemas claramente definidos
- Fácil de explicar (interpretável)

### IA Conectivista
Inspirada no funcionamento das **redes neurais biológicas**.

- Utiliza neurônios artificiais interconectados
- Aprende representações complexas a partir de dados
- Base das redes neurais e do **deep learning**

### Comparação

| Aspecto | Simbólica | Conectivista |
|---------|-----------|-------------|
| Base | Regras manuais | Dados + aprendizado |
| Interpretabilidade | Alta | Baixa |
| Flexibilidade | Baixa | Alta |
| Exemplo | Sistemas especialistas | Redes neurais, GPT |

# PARTE 3: Machine Learning — A Base de Tudo

## 📈 O que é Machine Learning?

**ML** ensina máquinas a reconhecer **padrões nos dados** ao invés de seguir regras escritas.

```
DADOS (exemplos) → ALGORITMO (aprende) → MODELO (prediz coisas novas)

Exemplo:
  Dados: 1000 emails com label "SPAM" ou "LEGÍTIMO"
  Algoritmo: Naive Bayes, SVM, Forest...
  Modelo: dado um email novo → prediz SPAM ou LEGÍTIMO? ✅
```

## 🎓 Tipos de Aprendizado

| Tipo | O que é? | Exemplo |
|------|----------|--------|
| **Supervisionado** | Dados com rótulos | Email → SPAM/LEGÍTIMO |
| **Não-supervisionado** | Dados sem rótulos | Agrupar clientes |
| **Reforço** | Aprender por recompensas | Robô aprende a andar |

### Aprendizado Supervisionado
Algoritmos aprendem a partir de exemplos rotulados (X → Y). O objetivo é predizer corretamente o output para exemplos nunca vistos.

### Aprendizado Não-supervisionado
O algoritmo identifica padrões e agrupamentos nos dados sem labels prévios. Clustering é o exemplo clássico.

### Aprendizado por Reforço
Um agente interage com um ambiente, recebendo recompensas/punições para aprender estratégias ótimas.

## 🧠 Classificação vs Regressão

```
CLASSIFICAÇÃO:
 Output: Categoria (valores discretos)
 Exemplo: Spam/Legítimo, Positivo/Negativo
 Métrica: Acurácia, Precision, Recall

REGRESSÃO:
 Output: Número contínuo
 Exemplo: Preço, Temperatura
 Métrica: MAE, RMSE
```

# PARTE 4: Fundamentos Matemáticos

## 𝛴 Conceitos Essenciais

### Vetores e Matrizes
Para que computadores processem dados (imagens, sons, textos), é necessário transformá-los em **representações numéricas**:
- **Vetores**: listas de números (ex: uma palavra como embedding)
- **Matrizes**: tabelas de números (ex: uma imagem como pixels)

### Funções de Ativação
Em redes neurais, funções como **ReLU** e **sigmoid** determinam se um "neurônio" será ativado:

```
ReLU(x)    = max(0, x)        → Simples e eficiente
Sigmoid(x) = 1 / (1 + e^(-x)) → Comprime output entre 0 e 1
```

Elas trazem **não-linearidade** às redes, permitindo que aprendam padrões complexos.

# PARTE 5: Importância dos Dados

## 📊 Qualidade dos Dados

A **qualidade**, **quantidade** e **diversidade** dos dados são essenciais para modelos robustos.

### Overfitting vs Underfitting

```
OVERFITTING (decorou demais):
 - Modelo aprende ruído e exceções do treino
 - Generaliza MAL para dados novos
 - Acurácia treino: 99%  |  Teste: 60%

UNDERFITTING (aprendeu de menos):
 - Modelo é simples demais
 - Não captura padrões reais
 - Acurácia treino: 55%  |  Teste: 50%

IDEAL:
 - Acurácia treino: 90%  |  Teste: 85%
```

### Como Evitar
- **Regularização**: limitar o ajuste ao treino
- **Validação cruzada**: testar em dados diferentes
- **Mais dados**: mais exemplos = melhor generalização

# PARTE 6: Naive Bayes — Primeiro Classificador (versão mais didática)

## 1. Ideia geral do Naive Bayes

Naive Bayes é um **classificador probabilístico**: ele decide a classe de um exemplo calculando **probabilidades** usando o **Teorema de Bayes**. [file:1]  

O apelido **“Naive” (ingênuo)** vem da suposição forte que ele faz:

> As **features são independentes entre si**, dado a classe.  

Em texto, isso significa assumir que a probabilidade de ver as palavras `"ganhe"` e `"dinheiro"` juntas em um email spam é o produto das probabilidades individuais de cada uma, **como se uma não influenciasse a outra**. [file:1]

Mesmo essa suposição sendo falsa na prática, o modelo costuma funcionar **surpreendentemente bem**, principalmente em problemas de texto como spam/sentimento. [file:1]

---

## 2. Como o modelo pensa (intuição)

Para cada classe possível (por exemplo, `SPAM` e `LEGÍTIMO`), o modelo avalia:

\[
P(\text{classe} \mid \text{palavras do email})
\]

Pelo Teorema de Bayes, isso é proporcional a:

\[
P(\text{classe}) \times P(\text{palavras} \mid \text{classe})
\]

- \(P(\text{classe})\): **probabilidade a priori** (por exemplo, qual fração dos emails do treino é spam). [file:1]  
- \(P(\text{palavras} \mid \text{classe})\): quão provável é ver aquelas palavras em um email daquela classe. [file:1]  

A parte “Naive” entra aqui:

\[
P(\text{palavras} \mid \text{classe}) \approx
\prod_i P(\text{palavra}_i \mid \text{classe})
\]

Ou seja, multiplica as probabilidades de cada palavra **individualmente**, assumindo independência.

O modelo escolhe a classe com maior valor:

\[
\text{classe\_predita} = \arg\max_{\text{classe}} P(\text{classe}) \prod_i P(\text{palavra}_i \mid \text{classe})
\]

---

## 3. Exemplo passo a passo (email de spam)

Imagine um email com as palavras `["ganhe", "dinheiro", "agora"]`. [file:1]

### Passo 1: Probabilidades das classes (priors)

Suponha que no conjunto de treino:

- 30 de 100 emails são `SPAM` → \(P(\text{SPAM}) = 0{,}3\)  
- 70 de 100 são `LEGÍTIMO` → \(P(\text{LEGÍTIMO}) = 0{,}7\)  

### Passo 2: Probabilidades das palavras dado SPAM

A partir das contagens nos emails marcados como SPAM, estimamos: [file:1]

- \(P(\text{"ganhe"} \mid \text{SPAM}) = 0{,}83\)  
- \(P(\text{"dinheiro"} \mid \text{SPAM}) = 0{,}80\)  
- \(P(\text{"agora"} \mid \text{SPAM}) = 0{,}67\)  

### Passo 3: Probabilidades das palavras dado LEGÍTIMO

Nos emails legítimos, estimamos: [file:1]

- \(P(\text{"ganhe"} \mid \text{LEGÍTIMO}) = 0{,}03\)  
- \(P(\text{"dinheiro"} \mid \text{LEGÍTIMO}) = 0{,}04\)  
- \(P(\text{"agora"} \mid \text{LEGÍTIMO}) = 0{,}14\)  

### Passo 4: Combinar tudo com Bayes

Para SPAM:

\[
P(\text{SPAM} \mid \text{palavras}) \propto
0{,}83 \times 0{,}80 \times 0{,}67 \times 0{,}3
\approx 0{,}133
\]

Para LEGÍTIMO:

\[
P(\text{LEGÍTIMO} \mid \text{palavras}) \propto
0{,}03 \times 0{,}04 \times 0{,}14 \times 0{,}7
\approx 0{,}000012
\]

### Passo 5: Decisão

Como \(0{,}133 > 0{,}000012\), o modelo classifica o email como:

\[
\boxed{\text{SPAM}}
\]

---

## 4. Vantagens e limitações

**Vantagens:** [file:1]

- **Muito rápido** de treinar e de prever.  
- Funciona bem com **poucos dados**.  
- Fácil de implementar e entender.  
- Bom como **baseline** em muitos problemas de texto.

**Limitações:** [file:1]

- A suposição de independência entre features é **irrealista**, principalmente em linguagem natural (palavras claramente se influenciam).  
- Pode ser **sensível a palavras novas** que não apareceram no treino (probabilidades muito baixas, precisa de suavização).

````python
# =============================================
# DEMONSTRAÇÃO: CLASSIFICAÇÃO DE SENTIMENTOS COM NAIVE BAYES
# =============================================

from sklearn.naive_bayes import MultinomialNB              # Algoritmo Naive Bayes para texto
from sklearn.feature_extraction.text import CountVectorizer # Converte texto em vetores de contagem de palavras
from sklearn.model_selection import train_test_split        # Separa treino e teste
from sklearn.metrics import accuracy_score                  # Mede acurácia do modelo

print("=" * 80)
print("🤖 DEMONSTRAÇÃO: NAIVE BAYES PARA CLASSIFICAÇÃO DE SENTIMENTOS")
print("=" * 80)

# -------------------------------------------------
# 1) Definição do dataset (texto + rótulo)
# -------------------------------------------------
avaliacoes_rotuladas = [
    ("Adorei o produto! Excelente qualidade!", "Positivo"),
    ("Produto muito bom, recomendo!", "Positivo"),
    ("Não gostei. Produto péssimo.", "Negativo"),
    ("Horrível, pior compra da minha vida", "Negativo"),
    ("Qualidade excelente, muito bom!", "Positivo"),
    ("Decepcionante, não recomendo", "Negativo"),
]

# Separa a lista em duas: textos e rótulos
textos_avaliacoes = [texto for texto, rotulo in avaliacoes_rotuladas]
rotulos_sentimento = [rotulo for texto, rotulo in avaliacoes_rotuladas]

print("\n📚 DATASET DE TREINO:")
for texto, rotulo in avaliacoes_rotuladas:
    emoji = "😊" if rotulo == "Positivo" else "😞"
    print(f"   {emoji} [{rotulo}] {texto}")

# -------------------------------------------------
# 2) Vetorização: texto -> matriz de contagens
# -------------------------------------------------
# CountVectorizer cria um vocabulário de palavras e transforma cada texto
# em um vetor de contagens (Bag of Words)
vetorizador_bow = CountVectorizer(lowercase=True)

# fit_transform:
#  - fit: aprende o vocabulário a partir dos textos
#  - transform: gera a matriz de features (documento x palavra)
matriz_features = vetorizador_bow.fit_transform(textos_avaliacoes)

# -------------------------------------------------
# 3) Separar em treino e teste
# -------------------------------------------------
X_treino, X_teste, y_treino, y_teste = train_test_split(
    matriz_features,
    rotulos_sentimento,
    test_size=0.33,      # 33% dos dados vão para teste
    random_state=42      # semente fixa para reprodutibilidade
)

# -------------------------------------------------
# 4) Treinar modelo Naive Bayes
# -------------------------------------------------
modelo_naive_bayes = MultinomialNB()   # Modelo para dados de contagem (texto)
modelo_naive_bayes.fit(X_treino, y_treino)  # Aprende a partir dos dados de treino

print("\n✅ Modelo treinado!")

# -------------------------------------------------
# 5) Avaliar desempenho (acurácia)
# -------------------------------------------------
acuracia_treino = accuracy_score(y_treino, modelo_naive_bayes.predict(X_treino))
acuracia_teste = accuracy_score(y_teste, modelo_naive_bayes.predict(X_teste))

print("\n📊 ACURÁCIA:")
print(f"   Treino: {acuracia_treino:.2%}")
print(f"   Teste:  {acuracia_teste:.2%}")

# -------------------------------------------------
# 6) Testar o modelo em novos textos
# -------------------------------------------------
print("\n🧪 TESTANDO EM NOVAS AVALIAÇÕES:")

novas_avaliacoes = [
    "Produto maravilhoso!",
    "Que decepção, muito ruim"
]

for texto_novo in novas_avaliacoes:
    # transform: usa o mesmo vocabulário aprendido para o novo texto
    features_novo = vetorizador_bow.transform([texto_novo])

    # predict: devolve o rótulo previsto (Positivo / Negativo)
    rotulo_previsto = modelo_naive_bayes.predict(features_novo)[0]

    # predict_proba: devolve as probabilidades para cada classe
    probabilidades = modelo_naive_bayes.predict_proba(features_novo)[0]

    print(f"\n   Texto: '{texto_novo}'")
    print(f"   Predição: {rotulo_previsto}")
    print(f"   Probabilidades: Positivo={probabilidades[1]:.2%}, Negativo={probabilidades[0]:.2%}")
````

# PARTE 7: Logistic Regression (versão mais didática)

## 1. Por que usar Logistic Regression?

- **Naive Bayes** supõe que todas as palavras (features) são independentes entre si, o que raramente é verdade em texto, mas é simples e rápido. [file:1]  
- **Logistic Regression** não faz essa suposição de independência: ela aprende um **peso diferente para cada feature** e combina tudo em uma única equação, ficando mais flexível e ainda interpretável. [file:1]  

Intuição prática:

- No **Naive Bayes**, cada palavra “vota” quase sozinha.  
- Na **Logistic Regression**, o modelo aprende uma equação que junta todos esses “votos” para decidir se o texto é **Positivo (1)** ou **Negativo (0)**.

---

## 2. Saída precisa ser probabilidade (0 a 1)

Em classificação binária (ex.: sentimento), queremos que o modelo diga coisas do tipo:

> “Esse review tem **0.87 de chance** de ser positivo.”

Para isso, a Logistic Regression funciona em **duas etapas**.

### 2.1. Modelo linear

Primeiro, o modelo faz uma combinação linear das features:

\[
z = w_1 x_1 + w_2 x_2 + \dots + w_n x_n + b
\]

- \(x_i\): features (por exemplo, valores da BoW ou TF–IDF para cada palavra). [file:1]  
- \(w_i\): pesos que o modelo aprende para cada feature. [file:1]  
- \(b\): bias (termo de ajuste). [file:1]  

Isso é igualzinho a uma regressão linear normal.

### 2.2. Função sigmoid

Depois, transforma esse \(z\) em uma **probabilidade entre 0 e 1** com a função sigmoid:

\[
\text{sigmoid}(z) = \frac{1}{1 + e^{-z}}
\]

Propriedades importantes: [file:1]

- Entrada: qualquer número real (\(-\infty\) até \(+\infty\)).  
- Saída: sempre entre 0 e 1 (perfeito para probabilidade).  

Comportamento:

- \(z \to -\infty\) → sigmoid \(\to 0\) (quase certeza de classe 0, por exemplo “Negativo”).  
- \(z = 0\) → sigmoid \(= 0{,}5\) (50/50, indeciso).  
- \(z \to +\infty\) → sigmoid \(\to 1\) (quase certeza de classe 1, por exemplo “Positivo”).  

No fim, a saída é:

\[
P(y = 1 \mid x) = \text{sigmoid}(z)
\]

Ou seja, a **probabilidade** do exemplo pertencer à classe 1. [file:1]

---

## 3. Como o modelo aprende os pesos?

Para aprender os melhores pesos \(w\) e bias \(b\), a Logistic Regression minimiza uma função de erro chamada **cross‑entropy loss**. [file:1]

### 3.1. Cross‑Entropy Loss

Para cada exemplo, com rótulo verdadeiro \(y \in \{0,1\}\) e previsão \(\hat y\) (probabilidade), o erro é:

\[
\text{Loss} = -y \log(\hat y) - (1-y)\log(1-\hat y)
\]

- Se \(y = 1\) e \(\hat y\) é alto (perto de 1), o erro é pequeno.  
- Se \(y = 1\) e \(\hat y\) é baixo (perto de 0), o erro explode (log de número pequeno).  

Isso força o modelo a dar alta probabilidade para a classe correta.

### 3.2. Gradient Descent

Para minimizar a Loss, usamos **gradient descent**: [file:1]

1. **Inicialização:** começa com pesos \(w\) e bias \(b\) aleatórios.  
2. **Cálculo da Loss:** calcula a Loss média em um conjunto de exemplos.  
3. **Gradiente:** calcula o gradiente da Loss em relação a cada \(w_i\) e \(b\) (direção em que o erro aumenta).  
4. **Atualização:** atualiza pesos na direção contrária ao gradiente:

   \[
   w \leftarrow w - \text{lr} \cdot \frac{\partial \text{Loss}}{\partial w}
   \]

   onde **lr** é a taxa de aprendizado.  
5. **Repetição:** repete esse processo até que a Loss pare de diminuir (convergência).

Assim o modelo encontra pesos que deixam as previsões o mais corretas possível no treino.

---

## 4. Vantagens e limitações

**Vantagens:** [file:1]

- Não assume independência entre features.  
- Produz **probabilidades bem calibradas** para a classe positiva.  
- É **interpretável**: cada peso \(w_i\) indica o quanto a feature \(x_i\) puxa para classe 1 (positivo) ou para classe 0 (negativo).  

**Limitações:** [file:1]

- A fronteira de decisão é **linear** no espaço das features.  
- Se o problema exigir uma separação muito **não linear**, você precisa:
  - criar features melhores (engenharia de features), ou  
  - usar modelos mais complexos (árvores, redes neurais, etc.).

````python
# =============================================
# DEMONSTRAÇÃO: LOGISTIC REGRESSION PARA SENTIMENTO
# =============================================

from sklearn.linear_model import LogisticRegression          # Modelo de regressão logística
from sklearn.feature_extraction.text import TfidfVectorizer  # Texto -> vetores TF-IDF
from sklearn.model_selection import train_test_split         # Separar treino e teste

print("=" * 80)
print("📈 DEMONSTRAÇÃO: LOGISTIC REGRESSION (CLASSIFICAÇÃO DE SENTIMENTOS)")
print("=" * 80)

# -------------------------------------------------
# 1) Definição do dataset (texto + rótulo 0/1)
# -------------------------------------------------
avaliacoes_rotuladas = [
    ("Adorei! Produto excelente!", 1),
    ("Muito bom, recomendo", 1),
    ("Qualidade perfeita", 1),
    ("Não gostei, decepcionante", 0),
    ("Produto ruim, horrível", 0),
    ("Péssima qualidade", 0),
    ("Excelente custo-benefício", 1),
    ("Não funciona bem", 0),
    ("Maravilhoso, adorei demais", 1),
    ("Pior compra ever", 0),
]

textos_avaliacoes = [texto for texto, rotulo in avaliacoes_rotuladas]
rotulos_binarios = [rotulo for texto, rotulo in avaliacoes_rotuladas]

print("\n📚 DATASET:")
for texto, rotulo in avaliacoes_rotuladas:
    emoji = "😊" if rotulo == 1 else "😞"
    print(f"   {emoji} [{rotulo}] {texto}")

# -------------------------------------------------
# 2) Vetorização com TF-IDF
# -------------------------------------------------
vetorizador_tfidf = TfidfVectorizer(lowercase=True)
matriz_features = vetorizador_tfidf.fit_transform(textos_avaliacoes)

# -------------------------------------------------
# 3) Separar em treino e teste
# -------------------------------------------------
X_treino, X_teste, y_treino, y_teste = train_test_split(
    matriz_features,
    rotulos_binarios,
    test_size=0.3,       # 30% dos exemplos vão para teste
    random_state=42
)

# -------------------------------------------------
# 4) Treinar a Logistic Regression
# -------------------------------------------------
modelo_logistico = LogisticRegression(random_state=42)
modelo_logistico.fit(X_treino, y_treino)

print("\n✅ Modelo Logistic Regression treinado!")

# -------------------------------------------------
# 5) Avaliar desempenho (acurácia)
# -------------------------------------------------
acuracia_treino = modelo_logistico.score(X_treino, y_treino)
acuracia_teste = modelo_logistico.score(X_teste, y_teste)

print("\n📊 ACURÁCIA:")
print(f"   Treino: {acuracia_treino:.2%}")
print(f"   Teste:  {acuracia_teste:.2%}")

# -------------------------------------------------
# 6) Inspecionar as palavras mais importantes
# -------------------------------------------------
print("\n🎯 TOP 5 PALAVRAS QUE MAIS PUXAM PARA POSITIVO:")

nomes_features = vetorizador_tfidf.get_feature_names_out()   # array de palavras
coeficientes = modelo_logistico.coef_[0]                     # pesos da classe positiva (1)

# segurança: garante que tamanhos batem
assert coeficientes.shape[0] == len(nomes_features), (
    f"Tamanhos diferentes: coef={coeficientes.shape[0]}, features={len(nomes_features)}"
)

indices_top5 = coeficientes.argsort()[-5:][::-1]             # índices dos 5 maiores pesos

for idx in indices_top5:
    palavra = nomes_features[idx]
    peso = coeficientes[idx]
    print(f"   '{palavra}': {peso:.4f}")

# -------------------------------------------------
# 7) Testar o modelo em um novo texto
# -------------------------------------------------
print("\n🧪 TESTANDO EM NOVO TEXTO:")

novo_texto = "Adorei muito, produto excelente!"
features_novo = vetorizador_tfidf.transform([novo_texto])

predicao_binaria = modelo_logistico.predict(features_novo)[0]
probabilidades = modelo_logistico.predict_proba(features_novo)[0]

rotulo_str = "Positivo" if predicao_binaria == 1 else "Negativo"

print(f"   Texto: '{novo_texto}'")
print(f"   Predição: {predicao_binaria} ({rotulo_str})")
print(f"   Probabilidade: Positivo={probabilidades[1]:.2%}, Negativo={probabilidades[0]:.2%}")
````

# PARTE 8: Limitações e Desafios da IA

## ⚡ Questões Importantes

### Viés de Dados
Dados de treino enviesados levam a decisões injustas pelos modelos de IA.
- Exemplo: modelo treinado apenas com dados de um grupo pode discriminar outros

### "Caixa Preta"
Muitas soluções modernas (redes profundas) têm **baixa interpretabilidade**, dificultando explicar por que deram uma resposta.

### Ética
Decisões automáticas (crédito, reconhecimento facial, contratação) levantam debates éticos relevantes.

### Privacidade
Dados pessoais usados para treinar IA podem expor usuários a riscos se não forem adequadamente protegidos.
